
# ReviewLens GPT — Training Environment

This notebook is used for the development and training
of ReviewLens GPT, an Indonesian Customer Review Insight Generator.

## Environment
- Platform: Kaggle Notebooks
- Framework: PyTorch
- GPU: NVIDIA Tesla T4
- CUDA: Enabled
- Dataset: Processed Dataset.csv

## Current Stage
Environment and dataset validation.

## Project Requirements
- Decoder-only Mini-Transformer
- Manual Multi-Head Self-Attention
- Causal Masking
- Next-token Prediction
- Random Weight Initialization
- No pretrained language models
- No external LLM inference APIs

## Dataset Policy
The original dataset is read-only.
All preprocessing outputs must be saved separately.


```python

import torch
import pandas as pd
from pathlib import Path

# 1. Check GPU
assert torch.cuda.is_available(), "GPU tidak tersedia!"

device = torch.device("cuda:0")

print("PyTorch:", torch.__version__)
print("CUDA:", torch.version.cuda)
print("GPU:", torch.cuda.get_device_name(0))
print("GPU Memory:",
      round(torch.cuda.get_device_properties(0).total_memory / 1024**3, 2),
      "GB")

# 2. Find dataset
csv_files = list(
    Path("/kaggle/input").rglob("Processed Dataset.csv")
)

assert csv_files, "Processed Dataset.csv tidak ditemukan!"

csv_path = csv_files[0]

# 3. Read header only, not customer data
columns = pd.read_csv(csv_path, nrows=0).columns.tolist()

required_columns = {
    "itemId",
    "reviewContent",
    "sentiment_label",
    "rating"
}

missing_columns = required_columns - set(columns)

assert not missing_columns, (
    f"Kolom wajib tidak ditemukan: {missing_columns}"
)

print("\nDataset berhasil ditemukan.")
print("Kolom wajib lengkap:", True)
print("Jumlah kolom:", len(columns))
print("Dataset siap digunakan.")
```
